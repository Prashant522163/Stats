Q1. Define Covariance and explain how it differs from Correlation in terms of scale and interpretation.
Covariance is a statistical measure that determines how two variables change together. It indicates whether two variables move in the same or opposite directions.

The formula for covariance is:
\[
\operatorname{Cov}(X,Y)=
\frac{\sum (X_i-\bar X)(Y_i-\bar Y)}{n}
\]

Where:
- \(X_i,Y_i\) = Individual observations.
- \(\bar X,\bar Y\) = Mean of X and Y.
- \(n\) = Number of observations.
Difference between Covariance and Correlation:
Covariance	Correlation
Measures how two variables change together.	Measures the strength and direction of their relationship.
Its value is not fixed.	Its value lies between -1 and +1.
Depends on the units and scale of variables.	It is unit-free.
Difficult to compare across datasets.	Easy to compare across datasets.

The correlation formula is:
\[
r=\frac{\operatorname{Cov}(X,Y)}{\sigma_X\sigma_Y}
\]

Q2. What does a positive, negative, and zero covariance indicate about the relationship between two variables?
Covariance can have three possible interpretations:

1. Positive Covariance: Both variables tend to move in the same direction. When X increases, Y also tends to increase.

Example: Height and weight.

2. Negative Covariance: Both variables tend to move in opposite directions. When X increases, Y tends to decrease.

Example: Product price and demand (in many situations).

3. Zero Covariance: There is no linear relationship between the two variables.

Example: Two variables that have no linear association.
Q3. Discuss the limitations of covariance as a measure of relationship between two variables. Why is correlation preferred in many cases?
Limitations of Covariance:
1. Covariance depends on the units of measurement.
2. Its value has no fixed range.
3. It is difficult to interpret the strength of the relationship.
4. It cannot easily be compared across different datasets.
5. Zero covariance does not necessarily mean that two variables are independent.
Why is Correlation Preferred?
1. Correlation always lies between -1 and +1.
2. It is independent of measurement units.
3. It clearly indicates the strength and direction of a relationship.
4. It is easier to understand and compare.
5. It provides standardized results.
Therefore, correlation is generally preferred because it gives a more meaningful and easily interpretable measure of the relationship between two variables.
Q4. Explain the difference between Pearson’s correlation coefficient and Spearman’s rank correlation coefficient. When would you prefer to use Spearman’s correlation?
Pearson's Correlation	Spearman's Correlation
Measures linear relationships.	Measures monotonic relationships.
Uses actual numerical values.	Uses ranks of values.
Suitable for continuous numerical data.	Suitable for ordinal or ranked data.
Sensitive to outliers.	Less sensitive to outliers.
Assumes a linear relationship.	Does not require a linear relationship.

Pearson's formula:
\[
r=\frac{\operatorname{Cov}(X,Y)}
{\sigma_X\sigma_Y}
\]

Spearman's formula:
\[
\rho=1-\frac{6\sum d^2}{n(n^2-1)}
\]

Where:
- \(d\) = Difference between the ranks.
- \(n\) = Number of observations.
When should we use Spearman's correlation?

Spearman's correlation is preferred when:
- Data is ordinal or ranked.
- Data contains significant outliers.
- The relationship is monotonic but not necessarily linear.
- Data does not satisfy the assumptions needed for Pearson's correlation.
Q5. If the correlation coefficient between two variables X and Y is 0.85, interpret this value in context. Can you infer causation from this value? Why or why not?
Given:
\[
r=0.85
\]

Interpretation:
- The correlation coefficient is positive.
- Its value is close to +1.
- Therefore, X and Y have a strong positive linear relationship.
- When X increases, Y generally tends to increase.
Can we infer causation?

No, we cannot conclude that X causes Y just because the correlation coefficient is 0.85.

Reasons:
1. Correlation does not imply causation.
2. Another variable may influence both X and Y.
3. The relationship may occur by coincidence.
Conclusion: There is a strong positive correlation between X and Y, but it does not necessarily mean that one causes the other.

Method to calculate covariance:

Step 1: Calculate the mean of X and Y.
\[
\bar X=\frac{\sum X}{n},\qquad
\bar Y=\frac{\sum Y}{n}
\]

Step 2: Calculate deviations from the means.
\[
(X_i-\bar X),\quad (Y_i-\bar Y)
\]

Step 3: Multiply the deviations.
\[
(X_i-\bar X)(Y_i-\bar Y)
\]

Step 4: Add all the products and divide by n.
\[
\boxed{\operatorname{Cov}(X,Y)=
\frac{\sum(X_i-\bar X)(Y_i-\bar Y)}{n}}
\]

Note: This formula calculates population covariance. For sample covariance, divide by \(n-1\).
